📌 Data Science Interview – Algorithm Practice List (Prioritized)

🔹 SQL Patterns You Should Recognize
```
	•	Top N per group → ROW_NUMBER + PARTITION BY
	•	Deduplication → ROW_NUMBER = 1
	•	Running totals → SUM() OVER
	•	Gaps & islands → LAG / LEAD
	•	Conditional counts → SUM(CASE WHEN ...)
	•	Anti-join → LEFT JOIN + IS NULL
```
