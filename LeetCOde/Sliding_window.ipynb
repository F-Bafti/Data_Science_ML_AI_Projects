# Longest Substring Without Repeating Characters

Given a string s, find the length of the longest substring without duplicate characters.



Example 1:

Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc", with the length of 3. Note that "bca" and "cab" are also correct answers.
Example 2:

Input: s = "bbbbb"
Output: 1
Explanation: The answer is "b", with the length of 1.
Example 3:

Input: s = "pwwkew"
Output: 3
Explanation: The answer is "wke", with the length of 3.
Notice that the answer must be a substring, "pwke" is a subsequence and not a substring.

```python
s = "au"
table = {} # ss:index
max_len = 1
L = 0
R = 1

for i, ss in enumerate(s):
    # print(i, ss)
    if ss in table:
        # print(ss, "already there", table[ss])
        table = {k:v for k, v in table.items() if v>=table[ss]}
        table[ss] = i
        L = i
        R = R + 1
        
    else:
        # print(ss, "not there")
        table[ss] = i
        R = R + 1
        
    current_len = len(table)
    max_len = max(max_len, current_len)

print(max_len)
    
```

```python
def lengthOfLongestSubstring(s):
    if len(s) == 0:
        return 0
    table = {} # ss:index
    max_len = 1
    L = 0
    R = 1

    for i, ss in enumerate(s):
        
        if ss in table:
            table = {k:v for k, v in table.items() if v>=table[ss]}
            table[ss] = i
            L = i
            R = R + 1
            
        else:
            table[ss] = i
            R = R + 1
        current_len = len(table)
        max_len = max(max_len, current_len)
    return max_len
```

Given a string s, find the length of the longest substring that contains at most TWO distinct characters.


```python
s = "eceba"

L = 0
R = 0
map = {}
max_len = 0

for i, ss in enumerate(s):
    if ss not in map:
        map[ss] = 1
        R = R + 1
        while len(map) > 2 :
            map[ss] = map[ss] - 1
            if map[ss] == 0:
                map.pop(ss)
            L = L + 1             
    else:
        map[ss] += 1
        R = R + 1
    max_len = max(max_len, R - L + 1)
```

# Minimum Size subarray sum

Given an array of positive integers nums and a positive integer target, return the minimal length of a subarray whose sum is greater than or equal to target. If there is no such subarray, return 0 instead.

 

Example 1:

Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
Explanation: The subarray [4,3] has the minimal length under the problem constraint.
Example 2:

Input: target = 4, nums = [1,4,4]
Output: 1
Example 3:

Input: target = 11, nums = [1,1,1,1,1,1,1,1]
Output: 0

```python
target = 7
nums = [2,3,1,2,4,3]
```

```python
res_len = []

for i in range(len(nums)):
    current_sum = 0
    current_len = 0  
    for j in range(i, len(nums)):
        current_sum = current_sum + nums[j]
        current_len = current_len + 1
        if current_sum >= target:
            res_len.append(current_len)
            break
print(min(res_len))
# But this is O(n^2)
```

```python
def minSubArrayLen(target, nums):
    L = 0
    total = 0
    res = float("inf")

    for R in range(len(nums)):
        total += nums[R]
        while total >= target:
            res = min(R - L + 1, res)
            total -= nums[L]
            L += 1
    if res == float("inf"):
        return 0
    else:
        return res
```

```python
minSubArrayLen(11, [1,1,1,1,1,1,1,1])
```

# Maximum Average Subarray I

You are given an integer array nums consisting of n elements, and an integer k.

Find a contiguous subarray whose length is equal to k that has the maximum average value and return this value. Any answer with a calculation error less than 10-5 will be accepted.

 

Example 1:

Input: nums = [1,12,-5,-6,50,3], k = 4
Output: 12.75000
Explanation: Maximum average is (12 - 5 - 6 + 50) / 4 = 51 / 4 = 12.75
Example 2:

Input: nums = [5], k = 1
Output: 5.00000

```python
nums = [1,12,-5,-6,50,3]
k = 4
```

```python
nums = [5]
k = 1
```

```python
if len(nums) == 1:
    res = nums[0]
else:
    res = -1 * float("inf")

for i in range(len(nums)):
    for j in range(i, len(nums)-k):
        avg = sum(nums[i + j : i + j + k]) / k 
        res = max(res, avg)

# The time complexity of this is potentially O(n^3)
```

```python
# Optimal

def findMaxAverage(nums, k):
    L = 0
    R = k
    total = sum(nums[L:R])
    res = total / k
    
    for i in range(len(nums) - k):
            total = (total - nums[L] + nums[R]) 
            res = max(res, total/k)
            R = R + 1
            L = L + 1
    return res
```
